```python
#Henbra de ADN inicial
hebra_ADN = "AGCCCTCCAGGACAGGCTGCATATGGCCCTGTGGATGCGCCTCCTGCCCCTGCGTGGCGCTGCTGGCCCTCTGGGGACCTGACCCAGCCGCAGCCTTTGTGAACCAACACCTGTGCGGCTCACACCTGGTGGAAGCTCTCTACCTAGTGTGCGGGGAACGAGGCTTCTTCTACACACCCAAGACCCGCCGGGAGGCAGAGGACCTGCAGGTGGGGCAGGTGGAGCTGGGCGGGGGCCCTGGTGCAGGCAGCCTGCAGCCCTTGGCCCTGGAGGGGTCCCTGCAGAAGCGTGGCATTGTGGAACAATGCTGTACCAGCATCTGCTCCCTCTACCAGCTGGAGAACTACTGCAACTGATCCTGACCCAGGTGTCTCAGGG"

#Transcripción
hebra_ARN = hebra_ADN.replace("T", "U")

# Aislo el gen deseado basandome en el codón de inicio y el de final del gen
inicio_gen = hebra_ARN.find("AUG")
final_gen = hebra_ARN.find("UGA") + 3

gen_aislado = hebra_ARN[inicio_gen:final_gen]

#Diccionario con la traducción de codones
codigo_genetico = {
    'UUU': 'Fen', 'UUC': 'Fen', 'UUA': 'Leu', 'UUG': 'Leu',
    'CUU': 'Leu', 'CUC': 'Leu', 'CUA': 'Leu', 'CUG': 'Leu',
    'AUU': 'Ile', 'AUC': 'Ile', 'AUA': 'Ile', 'AUG': 'Met',
    'GUU': 'Val', 'GUC': 'Val', 'GUA': 'Val', 'GUG': 'Val',
    'UCU': 'Ser', 'UCC': 'Ser', 'UCA': 'Ser', 'UCG': 'Ser',
    'CCU': 'Pro', 'CCC': 'Pro', 'CCA': 'Pro', 'CCG': 'Pro',
    'ACU': 'Thr', 'ACC': 'Thr', 'ACA': 'Thr', 'ACG': 'Thr',
    'GCU': 'Ala', 'GCC': 'Ala', 'GCA': 'Ala', 'GCG': 'Ala',
    'UAU': 'Tyr', 'UAC': 'Tyr', 'UAA': 'STOP', 'UAG': 'STOP',
    'CAU': 'His', 'CAC': 'His', 'CAA': 'Gln', 'CAG': 'Gln',
    'AAU': 'Asn', 'AAC': 'Asn', 'AAA': 'Lys', 'AAG': 'Lys',
    'GAU': 'Asp', 'GAC': 'Asp', 'GAA': 'Glu', 'GAG': 'Glu',
    'UGU': 'Cys', 'UGC': 'Cys', 'UGA': 'STOP', 'UGG': 'Trp',
    'CGU': 'Arg', 'CGC': 'Arg', 'CGA': 'Arg', 'CGG': 'Arg',
    'AGU': 'Ser', 'AGC': 'Ser', 'AGA': 'Arg', 'AGG': 'Arg',
    'GGU': 'Gly', 'GGC': 'Gly', 'GGA': 'Gly', 'GGG': 'Gly'
}

#Creo un buble para ensamblar la cadena de aminoacidos de la proteina
proteina = ""
for i in range(0, len(gen_aislado), 3):
    codon = gen_aislado[i:i+3]
    
    if len(codon) == 3:
        aminoacido = codigo_genetico[codon]
        proteina += aminoacido + "-"
    
    else:
        print("Aviso: Se ha descartado el fragmento incompleto", [codon], "al final de la secuencia.")

proteina = proteina.rstrip("-")
print(proteina)
```
